```python
'''Given two strings s and t, return true if s is a subsequence of t, or false otherwise.
A subsequence of a string is a new string that is formed from the original string by deleting some (can be none) of the characters without disturbing the relative positions of the remaining characters. 
(i.e., "ace" is a subsequence of "abcde" while "aec" is not).
Example 1:

Input: s = "abc", t = "ahbgdc"
Output: true'''
class Solution:
    def isSubsequence(self, s: str, t: str) -> bool:
        i = 0  # s

        for ch in t:
            if i < len(s) and s[i] == ch:
                i +=1
            
        return i == len(s)
```

```python
# Move Zeros :  maintaine the relative positon of the given array only move zeros at end
class Solution:
    def moveZeros(self, nums: List[int]):
        k= 0

        for i in range(len(nums)):
            if nums[i] != 0:
                nums[k], nums[i] = nums[i], nums[k]
                k +=1


nums = list(map(int, input('Enter the input array with white space: ').split()))
sol = Solution()
result = sol.moveZeros(nums)

print(f'zeros are moved from the given array: {nums} and maintained the relative position of the original array')
```

```python
# reverse a string 
class Solution:
    def reverseString(self, s:List[str]):

        left, right = 0, len(s)-1

        while left < right:
            s[left], s[right] = s[right], s[left]
            left += 1
            right -=1 

        return s

s = ["S","U","N","I","L"]
sol = Solution()
result = sol.reverseString(s)
print(s)
```

```python
class Solution:
    def reverseVowels(self, s: str) -> str:
        vowels = set('aeiouAEIOU')
        s = list(s)

        left, right = 0, len(s)-1

        while left<right:
            if s[left] not in vowels:
                left +=1
            elif s[right] not in vowels:
                right -=1
            else:
                s[left], s[right] = s[right], s[left]
                left += 1
                right -= 1
        
        # return  //o/p  will be  ['A','e' etc..]which a list but we needed str so used below return with join      
        return ''.join(s)
        
s = input('Enter a string with vowels: ' )
sol = Solution()
result = sol.reverseVowels(s)
print(result)
```

```python
#valid palindrome by removing alpha numeric character from the str ex: MA:M  --is a valid palindrome
class Solution:
    def isPalindrome(self, s: str) -> bool:
        left, right = 0, len(s)-1

        while left < right:
            # remove non-alphar char
            while left<right and not s[left].isalnum():
                left +=1
            while left<right and not s[right].isalnum():
                right -=1
            
            if s[left].lower() != s[right].lower():
                return False

            left += 1
            right -= 1    
            
        return True   
```

```python
''' You are given an integer array nums and an integer k.
In one operation, you can pick two numbers from the array whose sum equals k and remove them from the array.
Return the maximum number of operations you can perform on the array.
Example 1:

Input: nums = [1,2,3,4], k = 5
Output: 2
Explanation: Starting with nums = [1,2,3,4]:
- Remove numbers 1 and 4, then nums = [2,3]
- Remove numbers 2 and 3, then nums = []
There are no more pairs that sum up to 5, hence a total of 2 operations.'''
class Solution:
    def maxOperations(self, nums: List[int], k: int) -> int:
        nums.sort()
        left, right =0, len(nums)-1
        ops = 0

        while left < right:
            s = nums[left] + nums[right] 

            if s == k:
                ops +=1
                left +=1
                right -=1
            elif s < k:
                left +=1
            else: 
                right -=1
        return ops
```

```python
class Solution:
    def minPlatforms(self, arrival, departure):
        arrival.sort()
        departure.sort()
        

        plat_needed = 0
        max_platforms = 0

        i = 0  # pointer for arrival
        j = 0  # pointer for departure

        n = len(arrival) 

        while i < n and j < n:
            if arrival[i] <= departure[j]:
                plat_needed += 1
                max_platforms = max(max_platforms, plat_needed)
                i += 1
            else:
                plat_needed -= 1
                j += 1

        return max_platforms
    
# a = [900, 940, 950, 1100, 1500, 1800]
# d = [910, 1200, 1120, 1130, 1900, 2000]

# after sort 
a = [900, 940, 950, 1100, 1500, 1800]   #--i move i pointer until condition is ture and j stays the same
d = [910, 1120, 1130, 1200, 1900, 2000]   #j move j pointer when the condtion is false and i stays the same

sol = Solution()
result = sol.minPlatforms(a,d)
print(result)
```

```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
        
        dummy = ListNode(0)
        dummy.next = head

        slow  = dummy
        fast = dummy

        for _ in range(n):    #running loop whatever the n = 
            fast = fast.next  # fast pointer is two node ahead slow is at still at dummy

        while fast.next:        #run until the fast.next null 
            slow = slow.next    # slow pointer begins from here it moves on step forward
            fast = fast.next    # fast pointer moves one step forward from here till end

        # once the fast.next reached at end slow pointer skips the next node and return the points to the next.next node 
        slow.next = slow.next.next 

        return dummy.next
        
```
