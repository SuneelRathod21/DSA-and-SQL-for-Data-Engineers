```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        
        prev = None
        curr = head

        while curr:
            next_node = curr.next
            curr.next = prev
            prev = curr
            curr = next_node
        return prev
```

```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def removeNodes(self, head: Optional[ListNode]) -> Optional[ListNode]:
        def reverse(node):
            prev = None
            curr = node
            while curr:
                nxt = curr.next
                curr.next = prev
                prev = curr
                curr = nxt
            return prev

        head = reverse(head)

        max_val = float('-inf')
        curr = head
        pre = None

        while curr:
            if curr.val >= max_val:
                max_val = curr.val
                pre = curr
                curr = curr.next
            else:
                pre.next = curr.next
                curr = curr.next
        return reverse(head)
```

```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
        
        dummy = ListNode(0)
        dummy.next = head

        slow  = dummy
        fast = dummy

        for _ in range(n):    #running loop whatever the n = 
            fast = fast.next  # fast pointer is two node ahead slow is at still at dummy

        while fast.next:        #run until the fast.next null 
            slow = slow.next    # slow pointer begins from here it moves on step forward
            fast = fast.next    # fast pointer moves one step forward from here till end

        # once the fast.next reached at end slow pointer skips the next node and return the points to the next.next node 
        slow.next = slow.next.next 

        return dummy.next
        
```
